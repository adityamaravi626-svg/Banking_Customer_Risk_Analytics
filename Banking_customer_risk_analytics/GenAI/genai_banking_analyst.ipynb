```python
cleaned_df= pd.read_csv('/content/sample_data/bank_marketing_cleaned (2).csv')
```

```python
subscription_rate = (
    cleaned_df['y']
    .value_counts(normalize=True)
    .get('yes', 0) * 100
)
```

```python
print(
    f"Overall subscription rate: "
    f"{subscription_rate:.2f}%"
)
```

```python
contact_performance = (
    cleaned_df
    .groupby('contact')['y']
    .apply(
        lambda x: (x == 'yes').mean() * 100
    )
    .sort_values(ascending=False)
)
```

```python
contact_context = contact_performance.to_string()
```

```python
job_performance = (
    cleaned_df
    .groupby('job')['y']
    .apply(
        lambda x: (x == 'yes').mean() * 100
    )
    .sort_values(ascending=False)
)
```

```python
job_context = job_performance.to_string()
```
